```python
import numpy as np
from sklearn.cluster import MiniBatchKMeans
from sklearn.datasets import make_blobs

# Generate sample data
n_samples = 1500
random_state = 170
X, y = make_blobs(n_samples=n_samples, random_state=random_state)

# Number of clusters
k = 3

mb_kmeans = MiniBatchKMeans(n_clusters=k, random_state=random_state, n_init='auto')

# Fit the model to the data
mb_kmeans.fit(X)

# Get cluster labels for each data point
labels = mb_kmeans.labels_

# Get the cluster centers
cluster_centers = mb_kmeans.cluster_centers_

print("Cluster Labels (first 10):\n", labels[:10])
print("\nCluster Centers:\n", cluster_centers)
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

plt.figure(figsize=(10, 8))

# Plot data points colored by cluster labels
sns.scatterplot(x=X[:, 0], y=X[:, 1], hue=labels, palette='viridis', legend='full', s=50, alpha=0.7)

# Plot cluster centers
plt.scatter(cluster_centers[:, 0], cluster_centers[:, 1], marker='X', s=200, color='red', label='Cluster Centers', edgecolor='black')

plt.title('MiniBatchKMeans++ Clustering')
plt.xlabel('Feature 1')
plt.ylabel('Feature 2')
plt.legend()
plt.grid(True)
plt.show()
```
