```python
#Step 1: Import libraries
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from sklearn.preprocessing import StandardScaler
import umap
```

```python
#Step 2: Load Zomato dataset
df=pd.read_csv('/content/Zomato Restaurant Dataset.csv')

```

```python
#Step 3: Display dataset
print(df.head())
print(df.columns)
```

```python
#Step 4: Select numerical features
features = [
    'Average Cost for two',
    'Votes',
    'Aggregate rating',
    'Price range'
]

X=df[features].copy()
```

```python
#Step 5: Hande missing and infinite values
X=X.replace([np.inf,-np.inf],np.nan)
X=X.fillna(X.median())
```

```python
#Step 6: Feature scaling
scaler=StandardScaler()
X_scaled=scaler.fit_transform(X)
```

```python
#Step 7: Apply UMAP
reducer=umap.UMAP(n_components=2,min_dist=0.1,n_neighbors=15,metric='euclidean',random_state=42)
X_umapp=reducer.fit_transform(X_scaled)

```
